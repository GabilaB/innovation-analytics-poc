```python
# --------------------------------------------------
# CONFUSION MATRIX FIGURE
# --------------------------------------------------

from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

ROOT       = Path(".").resolve()
FIGURE_DIR = ROOT / "assets"
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

eval_df = pd.read_csv(ROOT / "evaluation" / "validation_evaluated.csv")

LABELS = [
    "innovation",
    "possible_innovation",
    "not_innovation",
    "insufficient_evidence",
]

SHORT = [
    "Innovation",
    "Possible",
    "Not",
    "Insufficient",
]

matrix = pd.crosstab(
    eval_df["human_innovation_presence"],
    eval_df["innovation_presence"],
).reindex(index=LABELS, columns=LABELS, fill_value=0)

fig, ax = plt.subplots(figsize=(7, 5))
im = ax.imshow(matrix.values, cmap="Blues", vmin=0)

ax.set_xticks(range(len(SHORT)))
ax.set_yticks(range(len(SHORT)))
ax.set_xticklabels(SHORT, fontsize=10)
ax.set_yticklabels(SHORT, fontsize=10)
ax.set_xlabel("Model prediction", fontsize=11)
ax.set_ylabel("Expert label", fontsize=11)
ax.set_title(
    "Confusion Matrix: Innovation Presence Classification",
    fontsize=12, pad=12
)

for i in range(len(LABELS)):
    for j in range(len(LABELS)):
        val   = matrix.values[i, j]
        color = "white" if val > matrix.values.max() * 0.6 else "black"
        ax.text(
            j, i, str(val),
            ha="center", va="center",
            fontsize=13, fontweight="bold", color=color,
        )

plt.colorbar(im, ax=ax, label="Record count")
plt.tight_layout()

save_path = FIGURE_DIR / "confusion_matrix.png"
plt.savefig(save_path, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", save_path)
```
